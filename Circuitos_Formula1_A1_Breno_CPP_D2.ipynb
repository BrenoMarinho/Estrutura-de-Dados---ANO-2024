````python
import pandas as pd #--> Importação da biblioteca pandas

# -- Criação de listas com os dados informados

CircuitName = ['Monza', 'Mônaco', 'Silverstone', 'Spa-Francorchamps', 'Gilles Villeneuve', 'Nurburgring', 'Interlagos', 'Hockenheimring', 'Suzuka', 'Zandvoort']
#-> Nome do circuito

TotalRuns = ['73', '69', '58', '56', '42', '41', '40', '37', '34', '33']
#-> Total de corridas

Location= ['Itália', 'Mônaco', 'Inglaterra', 'Bélgica', 'Canadá', 'Alemanha', 'Brasil', 'Hockenheimring', 'Japão', 'Holanda' ]
#-> Onde está localizada

TotalCurves = ['11', '19', '18', '19', '14', '16', '15', '15', '18', '14']
#-> Total de curvas da pista

WinningPilot = ['Michael Schumacher', 'Ayrton Senna', 'Lewis Hamilton', 'Michael Schumacher', 'Michael Schumacher', 'Michael Schumacher', 'Michael Schumacher', 'Michael Schumacher', 'Michael Schumacher', 'Lewis Hamilton']
#--> Piloto que mais ganhou no circuito

TrackSize = ['5,793 km', '3,337 km', '5,891 km', '7 km', '4,361 km', '5,148 km', '4,309 km', '3,629 km', '5,807 km', '4,3 km']
#-> Cumprimento da pista

Opening = ['1922', '1950', '1948', '1979', '1978', '1927','1940', '1932', '1962', '1948']
#-> Ano de inauguração do circuito

#Criação de um dicionário com as listas informadas
dados_circuitos= {
    'Nome do circuito': CircuitName,
    'Total de corridas': TotalRuns,
    'Localização': Location,
    'Total de Curvas': TotalCurves,
    'Piloto com mais vitória': WinningPilot,
    'Tamanho da pista': TrackSize,
    'Data de inauguração': Opening,

}

#Criação de um DataFrame com o dicionário
circuitos = pd.DataFrame(dados_circuitos)
````

> 1. Usando a biblioteca Pandas foi-se criado um DataFrame sobre os circuitos mais relevantes da Fórmula 1.
2. Utilizando listas para armazenar os dados coletados e, com elas, foi feito um dicionário, que foi utilizado para a criação do DataFrame e sua variável.


---

### Explicando o Código 🚨

* Piloto com mais vitória: Lista o nome do piloto com mais número de vitórias.
* Tamanho da pista:Lista do tamanho em quilômetro.
* Data de inauguração: Lista o ano em que o circuito doi inaugurado.

* O dicionário circuito é passado para o construtor  ```pd.DataFrame() ``` para criar um DataFrame chamado  ```dados_circuito ```.

### 7 - Adicione um item (linha) na sua base de dados.

````python
print("7) Adicione um item (linha) na sua base de dados.")
#Nessa fase dicionei uma nova linha no DataFrame com a ajuda do dicionario.
#Obs.: Foi um circuito aleatório
novo_circuito = {
    'Nome do circuito': 'Catalunha',
    'Total de corridas': '29',
    'Localização': 'Espanha',
    'Total de Curvas': '14',
    'Piloto com mais vitória': 'Valentino Rossi',
    'Tamanho da pista': '4,657 km',
    'Data de inauguração': '1991',
    }
circuitos = pd.concat([circuitos, pd.DataFrame(novo_circuito, index=[5])],ignore_index=True)
print(circuitos)
print()
````

---

### Explicação do Código 🚨

> 1. Novo Item a ser Adicionado:

* Este trecho define um novo circuito que será adicionado ao DataFrame. O item é representado como um dicionário com chaves correspondentes aos nomes das colunas do DataFrame, citado novamente abaixo:

````python
novo_circuito = {
    'Nome do circuito': 'Catalunha',
    'Total de corridas': '29',
    'Localização': 'Espanha',
    'Total de Curvas': '14',
    'Piloto com mais vitória': 'Valentino Rossi',
    'Tamanho da pista': '4,657 km',
    'Data de inauguração': '1991',
    }
````

> 2. Adicionar o Novo Item ao DataFrame Usando `pd.concat`:

* Neste trecho, estamos adicionando um novo circuito `novo_circuito` ao final do DataFrame `circuitos` utilizando a função `pd.concat()`. O parâmetro `ignore_index=True` garante que os índices das linhas sejam redefinidos após a concatenação, para evitar índices duplicados.

````python
circuitos = pd.concat([circuitos, pd.DataFrame(novo_circuito, index=[5])],ignore_index=True)
````

### 8 - Transponha a coluna para a linha em sua base de dados.

````python
print("8) Transponha a coluna para a linha em sua base de dados.")
# Transpor uma coluna específica, por exemplo, a coluna 'Localização'
coluna_localizacao_transposta = circuitos['Localização'].to_frame().T

# Exibir a coluna transposta
print()
print("\nColuna 'Localização' transposta para linha:")
display(coluna_localizacao_transposta)

# Transpor todo o DataFrame e exibi-lo
print("\nDataFrame transposto:")
circuitos_transposto = circuitos.T
circuitos_transposto
````


---

### Explicando o Código 🚨

> 1. Transpor uma Coluna Específica (Exemplo: Coluna 'Localização'):

* Aqui, estamos transpondo a coluna específica `'localização'` do DataFrame `circuitos`. Primeiro, selecionamos essa coluna utilizando `circuitos['Localização']`, depois convertemos em um DataFrame usando `to_frame()`, e finalmente aplicamos a transposição usando `.T`.

````python
coluna_localizacao_transposta = circuitos['Localização'].to_frame().T
````

> 2. Transpor Todo o DataFrame:

* Caso solictado, é possível transpor todo o DataFrame `circuitos` utilizando o método `.T`. Isso trocará as linhas pelas colunas e vice-versa.

````python
circuitos_transposto = circuitos.T
````

### 9 - Apresente em tela somente a 1ª e a 2ª coluna (rótulo) da base de dados.

````python
print("9) Apresente em tela somente a 1ª e a 2ª coluna (rótulo) da base de dados.")
#Utilizando uma variável do comando iloc, definindo as colunas que queremos alocar e apresentando-as em tela.
print(circuitos.loc[:, ['Nome do circuito', 'Total de corridas']])
print()
````


---

### Explicando o Código 🚨

> 1. Selecionar e Exibir Apenas a 1ª e 2ª Coluna:

* Aqui, estamos selecionando apenas as colunas `'Nome do circuito'` e `'Total de corridas'` do DataFrame `circuito` utilizando a notação de lista dupla `[:, ['Nome do circuito', 'Total de corridas']]` a partir da função `loc`, que é uma variação da função `iloc`.

````python
print(circuitos.loc[:, ['Nome do circuito', 'Total de corridas']])
````
