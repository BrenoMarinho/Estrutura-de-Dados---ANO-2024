```python
# 4) Construa um algoritmo que receba uma lista com "cinemas", solicite a um usuário que informa 3 tipos de filmes da
#    área da TI e apresenta quais foram os filmes

cinemas = ["cinema1", "cinema2", "cinema3"]
# type (cinemas)

filmes_de_ti = []
for i in range(3):
  filmes = input(f'informe o nome do {str(i+1)} primeiro filme da área de TI:')
  filmes_de_ti.append(filmes)

print(f'O comprimeiro da lista é: {len(filmes_de_ti)} o tipo de lista é {type(filmes_de_ti)}')



```

```python
# 5)Insira um elemento na lista de cinemas.

animal = ["vaca","boi","bezerro"]
inserir_animal = input (f'Insira mais um animal:')
animal.append(inserir_animal)
print(f'O novo animal inserido é {inserir_animal} \nTodos os elementos são:{animal}')

```

```python
# 6) Troque a posição dos elementos no índice [0] para [1]

animal.insert(0, animal.pop(1))
print(f'Os valores trocados foram: {animal}')
```
